```python
employee = {
    "name": "John",
    "email": "john@gmail.com",
    "department": "ICT"
}
```

```python
url="https://jsonplaceholder.typicode.com/users"
response = requests.post(url, json=employee)
```

```python
print(response.status_code)
```

```python
print(response.json())
```
